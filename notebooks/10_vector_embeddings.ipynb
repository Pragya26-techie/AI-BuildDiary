```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = ["Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"]
```

```python
from langchain_ollama import OllamaEmbeddings
```

```python
embeddings = OllamaEmbeddings(model="nomic-embed-text")
```

```python
vector = embeddings.embed_documents(documents)
```

```python
print(len(vector))
```

```python
len(vector[0])
```

#### storing vector embeddings value in vector_database

```python
from langchain_chroma import Chroma
```

```python
vector_store=Chroma.from_texts(texts=documents,
                               embedding=embeddings,
                               persist_directory ="./vector_db" )
```

```python
query = "Machine-learnings"
result = vector_store.similarity_search(query=query)
result
```

```python
query = "Data Science"
result = vector_store.similarity_search(query=query,k=2)
result
```
